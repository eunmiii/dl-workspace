## 3.2 고급 인덱싱 (Advanced Indexing)

```python
x = torch.tensor([[10, 20, 30],
                  [40, 50, 60],
                  [70, 80, 90]])

# 1. 불리언 마스킹 (50보다 큰 요소만 추출)
mask = x > 50
print("마스크 (Boolean Tensor):\n", mask)
print("50보다 큰 값만 추출:", x[mask])  # 1차원 텐서로 반환됨

# 2. 인덱스 리스트를 활용한 추출 (팬시 인덱싱)
# 0행과 2행 추출
rows = torch.tensor([0, 2])
print("\n0행과 2행 추출:\n", x[rows])

# 3. torch.index_select 사용 (dim=1: 열 기준 추출)
cols = torch.tensor([0, 2]) # 0열과 2열 선택
selected_cols = torch.index_select(x, dim=1, index=cols)        # dim=1 이 축의 번호를 나타내고 있음.
print("0열과 2열 추출 (index_select):\n", selected_cols)
```

stack은 차원의 차수가 추가되면서 붙음

```python
t1 = torch.tensor([[1, 2],
                   [3, 4]])

t2 = torch.tensor([[5, 6],
                   [7, 8]])

# 1. torch.cat (기존 차원 이어붙이기)
# dim=0: 세로 방향(행)으로 붙임
cat_dim0 = torch.cat([t1, t2], dim=0)
print("torch.cat (dim=0) Shape:", cat_dim0.shape)
print(cat_dim0)

# dim=1: 가로 방향(열)으로 붙임
cat_dim1 = torch.cat([t1, t2], dim=1)
print("\ntorch.cat (dim=1) Shape:", cat_dim1.shape)
print(cat_dim1)

# 2. torch.stack (새로운 차원 생성하며 쌓기)
# dim=0: 맨 앞에 새로운 차원 추가 (2, 2, 2)
stacked = torch.stack([t1, t2], dim=0)
print("\ntorch.stack (dim=0) Shape:", stacked.shape)
print(stacked)
```

## 3.4 텐서 분할 (Splitting Tensors)

chunk는 갯수, split 방향

```python
x = torch.arange(12).reshape(4, 3) # 4행 3열 텐서
print("분할할 원본 텐서 (4x3):\n", x)

# 1. torch.split: 행 방향(dim=0)으로 2행씩 분할
split_result = torch.split(x, split_size_or_sections=1, dim=0)      # 4로 하면 안 나눠짐. split은 몇ㅗ=행씩 나눠야함.  <>  chunk는 몇 등분
print("\nsplit (크기 2씩 분할 결과):")
for t in split_result:
    print(t)

# 2. torch.chunk: 행 방향(dim=0)으로 총 2개의 덩어리로 분할
chunk_result = torch.chunk(x, chunks=4, dim=0)
print("\nchunk (2개로 등분 결과):")
for t in chunk_result:
    print(t)
```

```python
x = torch.arange(12).reshape(4, 3) # 4행 3열 텐서
print("분할할 원본 텐서 (4x3):\n", x)

# 1. torch.split: 행 방향(dim=0)으로 2행씩 분할
split_result = torch.split(x, split_size_or_sections=2, dim=0)
print("\nsplit (크기 2씩 분할 결과):")
for t in split_result:
    print(t)

# 2. torch.chunk: 행 방향(dim=0)으로 총 2개의 덩어리로 분할
chunk_result = torch.chunk(x, chunks=2, dim=0)
print("\nchunk (2개로 등분 결과):")
for t in chunk_result:
    print(t)
```

4.3 데이터 타입 지정 및 형변환 (Type Casting)

```python
# 1. 생성 시 dtype 명시
x_float16 = torch.tensor([1.5, 2.5], dtype=torch.float16)
x_int32 = torch.zeros((2, 2), dtype=torch.int32)

# 2. 형변환 방식 1: .to() 메서드 (권장 표준 패턴)
# 디바이스(CPU/GPU) 변경과 dtype 변경을 동시에 수행할 수 있어 가장 많이 쓰입니다.
x_converted = x_float16.to(torch.float32)
print("to() 변환 후:", x_converted.dtype)

# 3. 형변환 방식 2: 축약 메서드 (.float(), .long(), .bool() 등)
x_long = x_float16.long()   # float16 -> int64 (소수점 버림 처리)
x_bool = x_float16.bool()   # 0이 아니면 True
print("long() 변환 후 (값):", x_long)
print("bool() 변환 후 (값):", x_bool)

# 4. 형변환 방식 3: 다른 텐서의 dtype 추종 (.type_as())
target_tensor = torch.ones(2, dtype=torch.float64)
x_matched = x_float16.type_as(target_tensor)
print("type_as() 변환 후:", x_matched.dtype)
```

## 4.4 자동 형변환 및 승격 (Type Promotion)

```python
int_tensor = torch.tensor([1, 2, 3], dtype=torch.int32)
float_tensor = torch.tensor([0.5, 1.5, 2.5], dtype=torch.float32)

# int32 + float32 -> float32로 자동 승격
result = int_tensor + float_tensor
print("연산 결과:", result)
print("자동 승격된 dtype:", result.dtype) # torch.float32
```

## 6.3 집계 연산 (Reduction Operations)

```python
x = torch.tensor([[1.0, 5.0, 3.0],
                  [4.0, 2.0, 6.0]]) # Shape: (2, 3)

# 1. 전체 요소 집계
print("전체 합:", x.sum().item())
print("전체 평균:", x.mean().item())

# 2. 차원(dim) 기준 집계
# dim=0: 행 방향(세로)으로 뭉쳐서 연산 -> 결과 Shape: (3,)
print("\n행 기준(dim=0) 평균:", x.mean(dim=0))

# dim=1: 열 방향(가로)으로 뭉쳐서 연산 -> 결과 Shape: (2,)
print("행 기준(dim=1) 최댓값:", x.max(dim=1).values)

# 3. keepdim=True 옵션
sum_keepdim = x.sum(dim=1, keepdim=True)
print("\nkeepdim=True 적용 합 Shape:", sum_keepdim.shape) # (2, 1)

# 4. Argmax (최댓값의 인덱스 위치)
print("\n행 기준 가장 큰 값의 인덱스 (dim=1):", x.argmax(dim=1)) # [1, 2]
```

## 6.4 제자리 연산 (In-place Operations)

```python
x = torch.tensor([1.0, 2.0, 3.0])
print("초기 메모리 주소:", id(x))

# Out-of-place 연산 (새로운 텐서 생성 -> 주소 변경됨)
x_new = x + 5.0
print("Out-of-place 후 x 주소:", id(x))
print("n_new 주소:", id(x_new))

# In-place 연산 (기존 텐서 메모리에 즉시 반영 -> 주소 동일)
x.add_(5.0)
print("In-place 연산 후 x 값:", x)
print("In-place 후 x 주소:", id(x)) # 초기 주소와 동일함

# 모든 요소를 0으로 초기화
x.zero_()
print("zero_() 적용 후:", x)
```
