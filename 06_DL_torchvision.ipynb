# Pytorch의 torchvision을 이용해 MNIST 데이터셋 로딩
https://docs.pytorch.org/vision/main/generated/torchvision.datasets.MNIST.html

```python
import numpy as np
import torchvision
import matplotlib.pyplot as plt

print(dir(torchvision.datasets))
```

```python
# PyTorch의 torchvision을 이용해 MNIST 데이터셋 로딩
train_dataset = torchvision.datasets.MNIST(root='./data', train=True, download=True)        # download를 처음할때는, False? or True?
test_dataset = torchvision.datasets.MNIST(root='./data', train=False, download=True)

print(f'train_dataset : {dir(train_dataset)}')
```

```python
# train_dataset.data 까지만 하면, 텐서로 가져옴.
print(f'train_dataset.data 타입 : {type(train_dataset.data)}')
print(f'train_dataset.data dir : {dir(train_dataset.data)}')
```

```python
# Keras 형태와 유사하게 NumPy ndarray로 변환

# 뒤에 .numpy() 지워도 잘 돌아감. 넘파이 지우면 => 텐서로 가져옴
# X_train = train_dataset.data.numpy()
# y_train = train_dataset.targets.numpy()
# X_test = test_dataset.data.numpy()
# y_test = test_dataset.targets.numpy()

X_train = train_dataset.data
y_train = train_dataset.targets
X_test = test_dataset.data
y_test = test_dataset.targets

print(X_train.shape) # (60000, 28, 28)
print(type(X_train)) # <class 'numpy.ndarray'>
#print(X_train[0])
#plt.imshow(X_train[0], cmap='gray')

print(X_test.shape)  # (10000, 28, 28)
print(type(X_train))
#print(X_test[0])
plt.imshow(X_test[0], cmap='gray')

plt.show()


# cmap='gray'를 생략하면 Matplotlib은 기본 컬러맵인 viridis를 적용합니다.
# 이 경우 흑백 이미지임에도 불구하고 화면에는 보라색, 초록색, 노란색이 섞인 화려한 색상으로 출력되어
# 데이터의 실제 모습을 직관적으로 확인하기 어렵다.
```

```python
# 시각화
plt.figure(figsize=(12, 4))  # 가로 12인치, 세로 4인치

for i in range(10):
    plt.subplot(2, 5, i+1)  # 2행 5열로 배치, i+1번째 위치
    plt.imshow(X_train[i], cmap='gray')  # 흑백 이미지
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')  # 축 제거

plt.tight_layout()
plt.show()
```
